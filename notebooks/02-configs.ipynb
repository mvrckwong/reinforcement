# Configs Usage

Examples of using the shared `configs` package from notebooks.

The `configs` package is now at the project root level, making it directly importable from anywhere in the project.


## Paths Configuration

Access local filesystem paths like project root, checkpoints, logs, and outputs directories.


```python
from configs.paths import get_paths, get_s3_paths

# Get local paths configuration
paths = get_paths()

print("Local Paths Configuration")
print("=" * 40)
print(f"Project Root:    {paths.project_root}")
print(f"Checkpoints:     {paths.checkpoints_dir}")
print(f"Logs:            {paths.logs_dir}")
print(f"Outputs:         {paths.outputs_dir}")
print(f"Configs:         {paths.configs_dir}")
print(f"Templates:       {paths.templates_dir}")
print(f"Env File:        {paths.env_file}")
```

## S3 Paths Configuration

Access S3 bucket names for cloud storage.


```python
# Get S3 paths configuration
s3_paths = get_s3_paths()

print("S3 Bucket Configuration")
print("=" * 40)
print(f"General:         {s3_paths.general_bucket}")
print(f"Artifacts:       {s3_paths.artifacts_bucket}")
print(f"Checkpoints:     {s3_paths.checkpoints_bucket}")
print(f"Logs:            {s3_paths.logs_bucket}")
```

## Run Context

Create a run context to correlate training artifacts (checkpoints, logs, metadata).


```python
from configs.run_context import RunContext

# Create a run context for a training run
ctx = RunContext(model_name="ppo_cartpole")

print("Run Context")
print("=" * 40)
print(f"Model Name:      {ctx.model_name}")
print(f"Run ID:          {ctx.run_id}")
print(f"Run Prefix:      {ctx.run_prefix}")
print()
print("S3 Paths for this run:")
print(f"  Checkpoints:   {ctx.checkpoints_prefix}")
print(f"  Best:          {ctx.checkpoint_best_prefix}")
print(f"  Latest:        {ctx.checkpoint_latest_prefix}")
print(f"  Logs:          {ctx.logs_prefix}")
print(f"  Metadata:      {ctx.metadata_key}")
```

## Logging Configuration

Access logging settings (level, retention, serialization).


```python
from configs.logging import get_logging_config

# Get logging configuration
log_config = get_logging_config()

print("Logging Configuration")
print("=" * 40)
print(f"Level:           {log_config.level}")
print(f"Retention:       {log_config.retention}")
print(f"Serialize:       {log_config.serialize}")
```

## Practical Example

Combine configs to set up a training run with proper output paths.


```python
# Example: Set up a training run with local checkpoint directory
from configs.paths import get_paths
from configs.run_context import RunContext

paths = get_paths()
ctx = RunContext(model_name="impala_trading")

# Create local checkpoint directory for this run
local_checkpoint_dir = paths.checkpoints_dir / ctx.run_prefix
local_checkpoint_dir.mkdir(parents=True, exist_ok=True)

print("Training Run Setup")
print("=" * 40)
print(f"Model:           {ctx.model_name}")
print(f"Run ID:          {ctx.run_id}")
print(f"Local Path:      {local_checkpoint_dir}")
print(f"S3 Prefix:       {ctx.checkpoints_prefix}")
print()
print(f"Directory created: {local_checkpoint_dir.exists()}")
```
